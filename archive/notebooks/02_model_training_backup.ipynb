# NFL Big Data Bowl 2026 - ST-GNN Transformer Training

This notebook trains a state-of-the-art Spatial-Temporal Graph Neural Network with Transformer encoder for NFL trajectory prediction.

## Model Architecture:
- **Node Encoder**: MLP encoding player features
- **Edge Encoder**: MLP encoding spatial relationships
- **Graph Attention Layers**: GATv2-style message passing
- **Evoformer Pairwise Updates**: AlphaFold-inspired pairwise interactions
- **Temporal Transformer**: Multi-head attention for temporal dynamics
- **Prediction Head**: Outputs (Δx, Δy) trajectory

## Training:
- Optimizer: AdamW
- Scheduler: Cosine Annealing
- Loss: MSE + acceleration smoothness
- Batch size: 32
- GPU: Auto-selection (optimized for dual RTX 5090)


```python
import sys
from pathlib import Path

# Add project root to path
project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm
import json
from collections import defaultdict
# Import config
import config

# Import preprocessing
from preprocessing import load_sequences

# Import model and utilities
from src.models.stgnn_transformer import STGNNRefine
from src.utils import NFLTrajectoryDataset, collate_fn

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"CUDA Version: {torch.version.cuda}")
    if torch.cuda.device_count() > 1:
        print(f"Multiple GPUs detected: {torch.cuda.device_count()}")
print()
```

## 1. Load Sequences


```python
print("Loading sequences...")
print("-" * 60)

sequences = load_sequences()

# Get unique game_ids for train/val split
game_ids = sorted(set(key[0] for key in sequences.keys()))
n_games = len(game_ids)
cutoff_idx = int(0.8 * n_games)
train_games = set(game_ids[:cutoff_idx])
val_games = set(game_ids[cutoff_idx:])

# Split sequences
train_sequences = {k: v for k, v in sequences.items() if k[0] in train_games}
val_sequences = {k: v for k, v in sequences.items() if k[0] in val_games}

print(f"Total sequences: {len(sequences):,}")
print(f"Train sequences: {len(train_sequences):,}")
print(f"Validation sequences: {len(val_sequences):,}")
print()
```

## 2. Load Input DataFrame for Graph Construction


```python
from preprocessing import load_all_train_inputs, normalize_play_direction, add_features

print("Loading input data for graph construction...")
print("-" * 60)

# Load and preprocess input data (needed for building graphs with all players)
input_df = load_all_train_inputs(config.DATA_PATH)
input_df = normalize_play_direction(input_df)
input_df = add_features(input_df)

print(f"✓ Loaded {len(input_df):,} input rows")
print(f"  Columns: {len(input_df.columns)}")
print()
```

## 3. Create DataLoaders


```python
# Dataset parameters
BATCH_SIZE = 32
MAX_TRAJECTORY_LENGTH = 100
MAX_PLAYERS = 22

print("Creating datasets...")
print("-" * 60)

train_dataset = NFLTrajectoryDataset(
    train_sequences,
    all_input_df=input_df,
    max_players=MAX_PLAYERS,
    max_trajectory_length=MAX_TRAJECTORY_LENGTH
)

val_dataset = NFLTrajectoryDataset(
    val_sequences,
    all_input_df=input_df,
    max_players=MAX_PLAYERS,
    max_trajectory_length=MAX_TRAJECTORY_LENGTH
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_fn,
    num_workers=4,
    pin_memory=True if torch.cuda.is_available() else False
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_fn,
    num_workers=4,
    pin_memory=True if torch.cuda.is_available() else False
)

print(f"Train batches: {len(train_loader)}")
print(f"Validation batches: {len(val_loader)}")
print()
```

## 4. Build Model


```python
# ---------------------------------------------------------
# 4. Model Hyperparameters
# ---------------------------------------------------------

HIDDEN_DIM = 256
NUM_GRAPH_LAYERS = 3
NUM_TEMPORAL_LAYERS = 4
NUM_HEADS = 8
DROPOUT = 0.1
NODE_INPUT_DIM = 13   # node features
EDGE_INPUT_DIM = 4    # edge features (not used in GAT but kept for consistency)

print("Building model...")
print("-" * 60)

# ---------------------------------------------------------
# Initialize NEW AlphaFold-inspired STGNN model
# ---------------------------------------------------------

model = STGNNRefine(
    node_dim=NODE_INPUT_DIM,
    hidden=HIDDEN_DIM,
    graph_layers=NUM_GRAPH_LAYERS,
    temporal_layers=NUM_TEMPORAL_LAYERS,
    heads=NUM_HEADS,
    max_len=100,
    dropout=DROPOUT,
).to(device)

print("Model loaded:", model.__class__.__name__)
print()

# ---------------------------------------------------------
# Parameter Count
# ---------------------------------------------------------
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print("-" * 60)
print()
```

## 5. Setup Training


```python
# ---------------------------------------------------------
# 5. Setup Training
# ---------------------------------------------------------

import torch.optim as optim
import math

# Training hyperparameters
LEARNING_RATE = 1e-4
WEIGHT_DECAY = 1e-5
NUM_EPOCHS = 15        # adjusted per your CTO decision
ACCELERATION_WEIGHT = 0.1

# ---------------------------------------------------------
# Optimizer
# ---------------------------------------------------------
optimizer = optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

# ---------------------------------------------------------
# Scheduler (warmup + cosine, AlphaFold-style)
# ---------------------------------------------------------
def build_scheduler(optimizer, warmup_steps=500, total_steps=20000):
    def lr_lambda(step):
        if step < warmup_steps:
            return float(step) / float(max(1, warmup_steps))
        progress = float(step - warmup_steps) / float(max(1, total_steps - warmup_steps))
        return 0.5 * (1.0 + math.cos(math.pi * progress))
    return optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)

scheduler = build_scheduler(optimizer)
current_step = 0

print("Optimizer and scheduler initialized.")
print("-" * 60)


# ---------------------------------------------------------
# compute_loss() stays EXACTLY as you have it
# ---------------------------------------------------------
def compute_loss(pred, target, mask):
    eps = 1e-8
    B, T, D = pred.shape

    # ----- Position MSE -----
    valid_pos = mask.sum()

    if valid_pos < 1:
        zero = torch.tensor(0.0, device=pred.device, requires_grad=True)
        return zero, zero, zero

    mse_num = ((pred - target)**2 * mask.unsqueeze(-1)).sum()
    mse_loss = mse_num / (valid_pos + eps)

    # ----- Acceleration smoothness -----
    if T < 3:
        return mse_loss, mse_loss, torch.tensor(0.0, device=pred.device)

    pred_vel = pred[:, 1:] - pred[:, :-1]
    target_vel = target[:, 1:] - target[:, :-1]

    pred_acc = pred_vel[:, 1:] - pred_vel[:, :-1]
    target_acc = target_vel[:, 1:] - target_vel[:, :-1]

    acc_mask = mask[:, 2:]
    valid_acc = acc_mask.sum()

    if valid_acc < 1:
        acc_loss = torch.tensor(0.0, device=pred.device)
    else:
        acc_num = ((pred_acc - target_acc)**2 * acc_mask.unsqueeze(-1)).sum()
        acc_loss = acc_num / (valid_acc + eps)

    total_loss = mse_loss + ACCELERATION_WEIGHT * acc_loss

    if not torch.isfinite(total_loss):
        print("⚠️ NaN/Inf in loss — using zero instead.")
        total_loss = torch.zeros(1, device=pred.device, requires_grad=True)

    return total_loss, mse_loss, acc_loss
```

## 6. Training Loop


```python
# ---------------------------------------------------------
# 6. TRAINING LOOP — FINAL, CLEAN, DEVICE-SAFE VERSION
# ---------------------------------------------------------

import torch

def safe_div(n, d, eps=1e-8):
    return n / (d + eps)

NUM_EPOCHS = 15  # keep in sync with Cell 5 if you change it

history = {
    'train_loss': [],
    'train_mse': [],
    'val_loss': [],
    'val_mse': [],
    'val_rmse': []
}

best_val_rmse = float('inf')
best_epoch = 0

print("Starting training...")
print("=" * 60)

for epoch in range(NUM_EPOCHS):

    # -----------------------
    # TRAINING PHASE
    # -----------------------
    model.train()
    train_losses = []
    train_mses = []

    train_pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [Train]")

    for batch in train_pbar:

        # ---- Move GraphFeatures to device ----
        graph = batch["graph"].to(device, non_blocking=True)

        # ---- Move targets & masks ----
        target = batch["target"].to(device, non_blocking=True)   # [B, T, 2]
        mask   = batch["mask"].to(device, non_blocking=True)     # [B, T]

        # ---- Initial position (absolute) ----
        if "initial_pos" in batch:
            initial_pos = batch["initial_pos"].to(device, non_blocking=True)  # [B, 2]
        else:
            # Use FIRST timestep of ground truth as starting position
            initial_pos = target[:, 0]  # [B, 2]

        optimizer.zero_grad(set_to_none=True)

        # ---- Forward (with initial position) ----
        pred = model(graph, initial_pos)   # [B, T, 2]

        # ---- Compute loss ----
        loss, mse_loss, acc_loss = compute_loss(pred, target, mask)

        if not torch.isfinite(loss):
            print("⚠️ NaN/Inf detected — skipping batch")
            continue

        # ---- Backward ----
        loss.backward()

        # ---- Gradient clipping ----
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)

        # ---- Update weights ----
        optimizer.step()

        # ---- LR Scheduler (PER BATCH) ----
        scheduler.step()

        train_losses.append(loss.item())
        train_mses.append(mse_loss.item())

        train_pbar.set_postfix({
            "loss": f"{loss.item():.4f}",
            "mse":  f"{mse_loss.item():.4f}"
        })

    # -----------------------
    # VALIDATION PHASE
    # -----------------------
    model.eval()
    val_losses = []
    val_mses = []
    val_rmses = []

    with torch.no_grad():
        val_pbar = tqdm(val_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [Val]")

        for batch in val_pbar:
            graph  = batch["graph"].to(device, non_blocking=True)
            target = batch["target"].to(device, non_blocking=True)
            mask   = batch["mask"].to(device, non_blocking=True)

            if "initial_pos" in batch:
                initial_pos = batch["initial_pos"].to(device, non_blocking=True)
            else:
                initial_pos = target[:, 0]

            pred = model(graph, initial_pos)

            loss, mse_loss, acc_loss = compute_loss(pred, target, mask)

            # RMSE from masked positions
            squared = (pred - target) ** 2 * mask.unsqueeze(-1)
            mse = safe_div(squared.sum(), mask.sum())
            rmse = torch.sqrt(mse)

            val_losses.append(loss.item())
            val_mses.append(mse_loss.item())
            val_rmses.append(rmse.item())

            val_pbar.set_postfix({
                "loss": f"{loss.item():.4f}",
                "rmse": f"{rmse.item():.4f}"
            })

    # -----------------------
    # EPOCH SUMMARY
    # -----------------------
    avg_train_loss = float(np.mean(train_losses)) if train_losses else float('nan')
    avg_train_mse  = float(np.mean(train_mses))   if train_mses  else float('nan')
    avg_val_loss   = float(np.mean(val_losses))   if val_losses  else float('nan')
    avg_val_mse    = float(np.mean(val_mses))     if val_mses    else float('nan')
    avg_val_rmse   = float(np.mean(val_rmses))    if val_rmses   else float('nan')

    history['train_loss'].append(avg_train_loss)
    history['train_mse'].append(avg_train_mse)
    history['val_loss'].append(avg_val_loss)
    history['val_mse'].append(avg_val_mse)
    history['val_rmse'].append(avg_val_rmse)

    lr_now = scheduler.get_last_lr()[0]

    print(f"\nEpoch {epoch+1}/{NUM_EPOCHS}")
    print(f"  Train Loss: {avg_train_loss:.4f} | Train MSE: {avg_train_mse:.4f}")
    print(f"  Val Loss:   {avg_val_loss:.4f} | Val MSE:   {avg_val_mse:.4f} | Val RMSE: {avg_val_rmse:.4f}")
    print(f"  LR: {lr_now:.6f}")

    # -----------------------
    # SAVE BEST MODEL
    # -----------------------
    if avg_val_rmse < best_val_rmse:
        best_val_rmse = avg_val_rmse
        best_epoch = epoch + 1

        save_path = config.PROJECT_ROOT / "models" / "stgnn_refine_best.pt"
        save_path.parent.mkdir(exist_ok=True)

        torch.save({
            'epoch': epoch+1,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'scheduler_state_dict': scheduler.state_dict(),
            'val_rmse': avg_val_rmse,
            'history': history
        }, save_path)

        print(f"  ✓ Saved best model (RMSE: {best_val_rmse:.4f})")

    print("-" * 60)

print("\nTraining complete!")
print(f"Best validation RMSE: {best_val_rmse:.4f} (epoch {best_epoch})")
```

```python
# Plot training curves
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Loss
axes[0].plot(history['train_loss'], label='Train Loss', linewidth=2)
axes[0].plot(history['val_loss'], label='Val Loss', linewidth=2)
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title('Training Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# MSE
axes[1].plot(history['train_mse'], label='Train MSE', linewidth=2)
axes[1].plot(history['val_mse'], label='Val MSE', linewidth=2)
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('MSE')
axes[1].set_title('Mean Squared Error')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# RMSE
axes[2].plot(history['val_rmse'], label='Val RMSE', linewidth=2, color='green')
axes[2].axhline(y=best_val_rmse, color='r', linestyle='--', label=f'Best: {best_val_rmse:.4f}')
axes[2].set_xlabel('Epoch')
axes[2].set_ylabel('RMSE')
axes[2].set_title('Validation RMSE')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------
# 7. Evaluation + Prediction Export (Final)
# ---------------------------------------------------------

print("\nEvaluating best model and exporting predictions...")
print("-" * 60)

# Load best model checkpoint
best_model_path = config.PROJECT_ROOT / "models" / "stgnn_refine_best.pt"
checkpoint = torch.load(best_model_path, map_location=device)
model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

all_predictions = []
all_keys = []

with torch.no_grad():
    eval_pbar = tqdm(val_loader, desc="Evaluating")

    for batch in eval_pbar:
        graph = batch["graph"].to(device)
        target = batch["target"].to(device)
        mask   = batch["mask"].to(device)
        keys   = batch["key"]   # (game_id, play_id, player_tracking_id)

        pred = model(graph)  # [B, T, 2]

        pred = pred.cpu().numpy()
        target = target.cpu().numpy()
        mask = mask.cpu().numpy()

        for i in range(len(keys)):
            seq_len = int(mask[i].sum())  # true trajectory length

            all_predictions.append({
                "key": keys[i],
                "pred": pred[i, :seq_len].tolist(),
                "true": target[i, :seq_len].tolist(),
            })
            all_keys.append(keys[i])

# Save predictions for external analysis
export_path = config.PROJECT_ROOT / "models" / "val_predictions.json"
import json

with open(export_path, "w") as f:
    json.dump(all_predictions, f, indent=2)

print(f"✓ Saved predictions to: {export_path}")
print("-" * 60)







# ---------------------------------------------------------
# 8. SAMPLE PREDICTIONS — FINAL, SAFE, STGNNRefine VERSION
# ---------------------------------------------------------

print("Loading best model checkpoint...")

model_path = config.PROJECT_ROOT / "models" / "stgnn_refine_best.pt"
checkpoint = torch.load(model_path, map_location=device)

model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

print("Model loaded. Generating sample predictions...")

num_samples = 4
sample_indices = np.random.choice(len(val_dataset), num_samples, replace=False)

fig, axes = plt.subplots(2, 2, figsize=(16, 16))
axes = axes.flatten()

with torch.no_grad():
    for idx, ax in zip(sample_indices, axes):

        sample = val_dataset[idx]

        graph_obj = sample["graph"]          # GraphFeatures object
        target = sample["target"]            # [T, 2]
        mask   = sample["mask"]              # [T]
        key    = sample["key"]
        traj_len = sample["trajectory_length"]

        # ---- Move GraphFeatures to device ----
        graph = graph_obj.to(device)

        # ---- Run model ----
        pred = model(graph)                  # [1, T, 2]
        pred = pred[0].cpu().numpy()         # → [T, 2]

        target = target.numpy()
        mask = mask.numpy()
        valid_mask = mask[:traj_len]

        # ---- Plot ground truth ----
        ax.plot(
            target[:traj_len, 0],
            target[:traj_len, 1],
            'b-o',
            label='True',
            markersize=4,
            linewidth=2,
            alpha=0.7
        )

        # ---- Plot predictions ----
        ax.plot(
            pred[:traj_len, 0],
            pred[:traj_len, 1],
            'r--s',
            label='Predicted',
            markersize=4,
            linewidth=2,
            alpha=0.7
        )

        # ---- Start point ----
        ax.plot(
            target[0, 0],
            target[0, 1],
            'go',
            markersize=10,
            label='Start',
            zorder=5
        )

        ax.set_xlabel("X Position (yards)")
        ax.set_ylabel("Y Position (yards)")
        ax.set_title(f"Game {key[0]}, Play {key[1]}, Player {key[2]}")
        ax.legend()
        ax.grid(True, alpha=0.3)
        ax.set_aspect('equal', adjustable='box')

plt.tight_layout()
plt.show()
```
